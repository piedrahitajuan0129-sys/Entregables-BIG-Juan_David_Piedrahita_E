```python
# Guardamos indicadores del estado inicial

duplicados_antes = df.duplicated().sum()

pesos_negativos_antes = (df["Weight kg"] < 0).sum()
longitudes_negativas_antes = (df["Body Length cm"] < 0).sum()

print("Duplicados antes:", duplicados_antes)
print("Pesos negativos antes:", pesos_negativos_antes)
print("Longitudes negativas antes:", longitudes_negativas_antes)
print("Tipo de fecha antes:", df["Observation date"].dtype)
```

```python
animal_type_map = {
    "red squirrel": "red squirrel",
    "red squirrell": "red squirrel",
    "red squirel": "red squirrel",
    "lynx": "lynx",
    "lynx?": "lynx",
    "European bison": "European bison",
    "European bison™": "European bison",
    "European bisson": "European bison",
    "European buster": "European bison",
    "hedgehog": "hedgehog",
    "ledgehod": "hedgehog",
    "wedgehod": "hedgehog"
}

df["Animal type"] = df["Animal type"].replace(animal_type_map)
```

```python
df["Country"] = df["Country"].replace({
    "PL": "Poland"
})
```

```python
df["Gender"] = df["Gender"].fillna("not determined")
```

```python
df["Observation date"] = pd.to_datetime(
    df["Observation date"],
    format="%d.%m.%Y",
    errors="coerce"
)
```

```python
df.loc[df["Weight kg"] < 0, "Weight kg"] = np.nan

df.loc[df["Body Length cm"] < 0, "Body Length cm"] = np.nan
```

```python
df = df.drop_duplicates()
```

```python
print("========== VALIDACIÓN DE LA LIMPIEZA ==========")

print("\n1. Duplicados:")
print("Antes:", duplicados_antes)
print("Después:", df.duplicated().sum())

print("\n2. Valores negativos:")
print("Pesos negativos después:", (df["Weight kg"] < 0).sum())
print("Longitudes negativas después:", (df["Body Length cm"] < 0).sum())

print("\n3. Tipo de fecha:")
print(df["Observation date"].dtype)

print("\n4. Tipos de animal:")
print(df["Animal type"].value_counts(dropna=False))

print("\n5. Países:")
print(df["Country"].value_counts(dropna=False))

print("\n6. Género:")
print(df["Gender"].value_counts(dropna=False))
```

Justificación de la limpieza: En esta etapa revisé y limpié los datos para corregir algunos errores que podían afectar el análisis. Se corrigieron nombres de animales y países que estaban escritos de diferentes formas, se trataron los datos faltantes de género, se convirtió la fecha a un formato adecuado y se corrigieron valores negativos que no tenían sentido para el peso y la longitud de los animales. También se eliminaron los registros duplicados para evitar contar la misma observación más de una vez. Los datos faltantes de algunas variables se mantuvieron cuando no era posible completar la información de forma confiable.

Punto 3. Crear una variable con np.where() "Variable longitud del Animal"

```python
mediana_longitud = np.nanmedian(df["Body Length cm"])

print("Mediana de longitud corporal:", mediana_longitud, "cm")
```

```python
df["Longitud corporal"] = np.where(
    df["Body Length cm"] >= mediana_longitud,
    "Igual o mayor a la mediana",
    "Menor a la mediana"
)
```

Función Estadistica, peso promedio del animal 

```python
promedio_peso = np.nanmean(df["Weight kg"])

print("Peso promedio de los animales:", round(promedio_peso, 2), "kg")
```

```python
mediana_peso = np.nanmedian(df["Weight kg"])

print("Promedio de peso:", round(promedio_peso, 2), "kg")
print("Mediana de peso:", round(mediana_peso, 2), "kg")
```

La mediana de la longitud corporal fue de 21 cm, por lo que se utilizó este valor como referencia para crear una nueva variable mediante np.where(), clasificando las observaciones entre animales con una longitud igual o mayor a la mediana y animales con una longitud menor. Además, utilizando np.nanmean() se obtuvo un peso promedio de aproximadamente 47,58 kg. Al compararlo con la mediana de peso de 0,35 kg, se observa una diferencia considerable, posiblemente relacionada con la presencia de animales de diferentes tamaños y pesos y con valores extremos en la variable.

4. Cuántos registros existen para cada tipo de animal.

```python
observaciones_animal = df.groupby("Animal type").size().sort_values(ascending=False)

print(observaciones_animal)
```

```python
promedio_peso_animal = (
    df.groupby("Animal type")["Weight kg"]
    .mean()
    .sort_values(ascending=False)
)

print(promedio_peso_animal)
```

¿Cuál es el promedio de una variable numérica según el país?

```python
promedio_peso_pais = (
    df.groupby("Country")["Weight kg"]
    .mean()
    .sort_values(ascending=False)
)

print(promedio_peso_pais)
```

cantidad de animales por país

```python
observaciones_pais = (
    df.groupby("Country")
    .size()
    .sort_values(ascending=False)
)

print(observaciones_pais)
```

En este punto utilicé groupby() para organizar la información y poder comparar los datos de acuerdo con el tipo de animal y el país. Se analizaron cuatro aspectos: la cantidad de observaciones por tipo de animal, el peso promedio por tipo de animal, el peso promedio por país y la cantidad de observaciones por país. Estos análisis permiten identificar qué animales tienen mayor presencia en los datos, comparar sus pesos promedio y observar cómo se distribuyen las observaciones entre los diferentes países. Esto facilita la interpretación de los datos y permite encontrar diferencias y patrones que no serían tan fáciles de identificar revisando los registros individualmente.
