1. ¿Qué solicita el ejercicio?
El taller solicita crear una lista llamada estudiantes. Cada elemento debe ser una tupla con la estructura (nombre, edad, nota) y debe haber mínimo 8 estudiantes.


```python
estudiantes = [
    ("Monica", 20, 4.5),
    ("Sebastian", 22, 3.8),
    ("Laura", 19, 4.2),
    ("Milton", 21, 2.9),
    ("Antonella", 20, 3.5),
    ("Juan", 23, 4.8),
    ("Sofía", 18, 3.9),
    ("Miguel", 22, 2.7)
]
```

2. ¿Qué solicita el ejercicio?
Se debe recorrer la lista con un ciclo for e imprimir cada estudiante con un formato similar al indicado en el taller.


```python
for estudiante in estudiantes:
    nombre, edad, nota = estudiante
    print(f"{nombre} tiene {edad} años y obtuvo una nota de {nota}")
```

3. ¿Qué solicita el ejercicio?
La clasificación indicada por el taller es: nota >= 4.5: Excelente; entre 4.0 y 4.49: Bueno; entre 3.0 y 3.99: Aceptable; menor de 3.0: Reprobó.


```python
for estudiante in estudiantes:
    nombre, edad, nota = estudiante

    if nota >= 4.5:
        clasificacion = "Excelente"
    elif nota >= 4.0:
        clasificacion = "Bueno"
    elif nota >= 3.0:
        clasificacion = "Aceptable"
    else:
        clasificacion = "Reprobó"

    print(f"{nombre}: {clasificacion}")
```

4. ¿Qué solicita el ejercicio?
Se debe calcular el promedio de todas las notas utilizando un ciclo y una variable acumuladora.


```python
suma_notas = 0

for estudiante in estudiantes:
    nombre, edad, nota = estudiante
    suma_notas += nota

promedio = suma_notas / len(estudiantes)

print(f"El promedio general es: {promedio:.2f}")
```

5.¿Qué solicita el ejercicio?
Se debe solicitar el nombre mediante input(). Si existe en la lista, se informa que fue encontrado; de lo contrario, se informa que no existe.


```python
nombre_buscar = input("Ingrese el nombre del estudiante: ")

encontrado = False

for estudiante in estudiantes:
    nombre, edad, nota = estudiante

    if nombre.lower() == nombre_buscar.lower():
        print(f"El estudiante {nombre} fue encontrado.")
        encontrado = True
        break

if not encontrado:
    print(f"No se encontró ningún estudiante con el nombre {nombre_buscar}.")
```

6. ¿Qué solicita el ejercicio?
La llave debe ser el nombre del estudiante y el valor, la ciudad donde vive. Después se recorre el diccionario.


```python
ciudades = {
    "Monica": "Bogotá",
    "Sebastian": "Medellín",
    "Laura": "Pereira",
    "Milton": "Cali",
    "Antonella": "Bogotá",
    "Juan": "Medellín",
    "Sofía": "Bogotá",
    "Miguel": "Cali"
}

for nombre, ciudad in ciudades.items():
    print(f"{nombre} vive en {ciudad}")
```

7.Qué solicita el ejercicio?
A partir del diccionario anterior se debe construir otro diccionario con el número de estudiantes por ciudad.


```python
cantidad_por_ciudad = {}

for ciudad in ciudades.values():

    if ciudad in cantidad_por_ciudad:
        cantidad_por_ciudad[ciudad] += 1
    else:
        cantidad_por_ciudad[ciudad] = 1

print(cantidad_por_ciudad)
```

8. El programa debe permitir ingresar números hasta que el usuario introduzca 0. El cero no se cuenta y no se suma. Al final se muestra cantidad, suma y promedio.


```python
cantidad = 0
suma = 0

while True:
    numero = float(input("Ingrese un número (0 para terminar): "))

    if numero == 0:
        break

    suma += numero
    cantidad += 1

if cantidad > 0:
    promedio = suma / cantidad
else:
    promedio = 0

print(f"Cantidad de números ingresados: {cantidad}")
print(f"Suma total: {suma}")
print(f"Promedio: {promedio:.2f}")
```

9. ¿Qué solicita el ejercicio?
Se deben recorrer los números del 1 al 30. Los múltiplos de 3 deben provocar continue y al llegar al número 25 se debe usar break.


```python
for numero in range(1, 31):

    if numero % 3 == 0:
        continue

    if numero == 25:
        break

    print(numero)
```
